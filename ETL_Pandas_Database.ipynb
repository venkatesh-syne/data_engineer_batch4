```python
import pandas as pd
from sqlalchemy import create_engine
from datetime import datetime
```

```python
# Configuration 

customers_df = pd.read_csv(r"E:\Python\customers.csv")

products_df = pd.read_excel(r"E:\Python\products.xlsx")

orders_df = pd.read_json(r"E:\Python\orders.json")
```

```python
# Logging
import logging
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s %(levelname)s %(message)s"
)
```

```python
customers_df.drop_duplicates(
    subset=["customer_id"],
    inplace=True
)

products_df.drop_duplicates(
    subset=["product_id"],
    inplace=True
)

orders_df.drop_duplicates(
    subset=["order_id"],
    inplace=True
)

customers_df.fillna(
    {
        "name": "UNKNOWN",
        "city": "UNKNOWN"
    },
    inplace=True
)

products_df.fillna(
    {
        "price": 0
    },
    inplace=True
)

orders_df.fillna(
    {
        "quantity": 0
    },
    inplace=True
)
```

```python
sales_df = pd.merge(
    orders_df,
    customers_df,
    on="customer_id",
    how="left"
)
sales_df.info()
```

```python
sales_df = pd.merge(
    sales_df,
    products_df,
    on="product_id",
    how="left"
)
sales_df.info()
```

```python
sales_df["sales_amount"] = (
    sales_df["quantity"]
    * sales_df["price"]
)

sales_df["name"] = (
    sales_df["name"]
    .str.upper()
)

sales_df["load_timestamp"] = (
    datetime.now()
)

sales_df["order_date"] = pd.to_datetime(
    sales_df["order_date"]
)

sales_df["year"] = (
    sales_df["order_date"]
    .dt.year
)

sales_df["month"] = (
    sales_df["order_date"]
    .dt.month
)
```

```python
print("\nNull Count")

print(
    sales_df.isnull().sum()
)

print("\nDuplicate Count")

print(
    sales_df.duplicated().sum()
)
```

```python
summary_df = (
    sales_df
    .groupby(
        ["customer_id", "name"]
    )["sales_amount"]
    .sum()
    .reset_index()
)
```
