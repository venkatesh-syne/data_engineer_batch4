```python
#Q2 count vowels in string

str1='This is Data Engineering Class'
count=0
for ch in str1:
    if ch in 'aeiou':
        count+=1
print(f'Vowe is count is {count}')
```

```python
#Q4 Check whether string is palindrome

str1='This is Data Engineering Class'
str2='madam'

str1_rev=str1[::-1]
str2_rev=str2[::-1]

if str1==str1_rev:
    print('string is palindrome')
else:
    print('string not palindrome')
    
if str2==str2_rev:
    print('string is palindrome')
```

```python
#Find the largest number in the list

list3=[3,4,60,6,7]
largest=list3[0]
for ele in list3:
    if ele>largest:
        largest=ele
print(f'largest element is {largest}')
     
```

```python
#Find the smallest number in the list

list3=[3,4,60,6,7]
smallest=list3[0]
for ele in list3:
    if ele<smallest:
        smallest=ele
print(f'largest element is {smallest}')
```

```python
#Find the sum of all elements in the list

list3=[3,4,60,6,7]
sum=0
for ele in list3:
    sum+=ele
print(sum)

```

## dictionaries

```python
#access value using key

d1={1:'a',2:'b',3:'c'}
print(d1[1])
```

```python
#add new key value pair
d1 = {1:'a', 2:'b', 3:'c'}

d1[4] = 'd'

print(d1)
```

```python
#update dictionary
d1 = {1:'a', 2:'b', 3:'c'}

d1[2] = 'z'

print(d1)
```

```python
#delete
d1 = {1:'a', 2:'b', 3:'c'}
del d1[1]
d1
```

```python
#loop through dictionary key and vales
d2 = {1:'a', 2:'b', 3:'c'}


for key, value in d2.items():
    print(key, value)
```

```python
#check if key exist

d1 = {1:'a', 2:'b', 3:'c'}

if 2 in d1:
    print("Key exists")
else:
    print("Key not found")
```

```python
d1 = {1:'a', 2:'b', 3:'c'}

print(d1.keys())
```
