```python
#find the largest among 3 numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Largest:", a)
elif b >= a and b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)
```

```python
#check whether a year is leap year
year = int(input("Enter year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not Leap Year")
```

```python
#simple calculator
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
op = input("Enter operator: ")

if op == "+":
    print(a + b)
elif op == "-":
    print(a - b)
elif op == "*":
    print(a * b)
elif op == "/":
    print(a / b)
else:
    print("Invalid operator")
```

##While loop

```python
#print numbers from 1 -20
i = 1

while i <= 20:
    print(i)
    i += 1
```

```python
#Sum of digits
num = int(input("Enter number: "))
total = 0

while num > 0:
    digit = num % 10
    total += digit
    num = num // 10

print("Sum =", total)
```

```python
#reverse a number
num=int(input('Enter a number'))
reverse=0

while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num=num//10
print(reverse)

```

```python
#check number is palindrom

n=int(input('Enter a number'))
original=n
rev=0

while n>0:
    digit=n%10
    rev=rev*10+digit
    n=n//10

if original==rev:
    print('palindrome')
else:
    print('no')
```

```python
#factorial of num
num = int(input("Enter number: "))

fact = 1
i = 1

while i <= num:
    fact = fact * i
    i += 1

print("Factorial =", fact)
```

```python
#fibonacci series

n = int(input("Enter number of terms: "))

a = 0
b = 1
count = 1

while count <= n:
    print(a)
    c = a + b
    a = b
    b = c
    count += 1


```

```python
#check whether a number is armstrong number

n=int(input('Enter a number'))
org=n
sum1=0

while n>0:
    digit=n%10
    sum1+=digit**3
    n=n//10

if sum1==org:
    print('Armstrong')
else:
    print('no')
```

```python
#find second largest element in a list

list1 = [1, 4, 6, 7, 34, 67]

largest = list1[0]
second = list1[0]

for ele in list1:
    if ele > largest:
        second = largest
        largest = ele

    elif ele > second and ele != largest:
        second = ele

print("Second largest:", second)



```

```python
#remove duplicates from list using loop

list1 = [1,2,2,3,4,4,4]
list2 = []

for ele in list1:
    if ele not in list2:
        list2.append(ele)

print(list2)


```

```python
#create login system with 3 attempts


correct_password = "admin123"
attempts = 1

while attempts <= 3:
    password = input("Enter password: ")

    if password == correct_password:
        print("Login successful")
        break

    else:
        print("Wrong password")
        attempts += 1

if attempts > 3:
    print("You are blocked")
```
