```python
import csv
import json
import logging
import sys
import os

logging.basicConfig(filename="app.log", level=logging.INFO)

def csv_to_json(input_file, output_file):
    '''
    This function is used to convert csv file to json
    reading csv file
    writing to json structure
    '''
    inp = input_file
    try:
        
        if not os.path.exists(inp):
            print(f"Input file not found: {inp}")
            return

        data = []

        with open(inp, "r",encoding = "utf-8") as csv_file:
            reader = csv.DictReader(csv_file)
            for row in reader:
                # print(row)
                data.append(row)
        # print(data)
        # return data

       
        os.makedirs(os.path.dirname(output_file), exist_ok=True)

        with open(output_file, "w") as json_file:
            json.dump(data, json_file, indent=2)

        logging.info("Conversion successful")
        print("Conversion successful")

    except Exception as e:
        logging.exception(e)
        print("Error:", e)


if _name_ == "_main_":
    print("Arguments received:", sys.argv)  # DEBUG

    
    if len(sys.argv) == 3:
        input_file = sys.argv[1]
        output_file = sys.argv[2]

        print(f"Input file: {input_file}")
        print(f"Output file: {output_file}")

        data1 = csv_to_json(input_file, output_file)
        print(data1)

    else:
        print("Usage:")
        print("python csv_to_json.py <input_csv> <output_json>")
```
