```python
#write a recursive function to calculate factorial

def calculate_factorial(n):
    if n==1:
        return 1
    else:
        return n * calculate_factorial(n-1)
```

```python
#Write a decorator to measure execution time of a function.

import time

def measure_time(func):
    def wrapper():
        start = time.time()

        func()

        end = time.time()

        print("Execution time:", end - start)

    return wrapper


@measure_time
def test():
    for i in range(1000000):
        pass

test()

```

```python
#Create a closure that remembers previous values (running total).

def running_total():
    total = 0

    def add(num):
        nonlocal total
        total += num
        return total

    return add


calc = running_total()

print(calc(10))   # 10
print(calc(5))    # 15
print(calc(20))   # 35
```

```python
#Write a generator function to yield even numbers up to n.

'''yield makes the function a generator
A generator returns values one at a time, instead of storing all values at once
Each time the loop runs, yield pauses the function and sends the next value back'''

def even_numbers(n):
    for i in range(0, n + 1):
        if i % 2 == 0:
            yield i


for num in even_numbers(10):
    print(num)
```

```python
# Create a function that returns another function (higher-order function).

'''
Higher-order function → function that takes or returns another function
return inner → returns the function itself
return inner() → executes the function and returns its value'''

def outer():
    def inner():
        return "Hello"

    return inner


result = outer()

print(result())
```

```python
#Write a decorator to check if user is authenticated before running a function.


def authenticate(func):
    def wrapper():
        is_authenticated = True

        if is_authenticated:
            func()
        else:
            print("Access denied")

    return wrapper


@authenticate
def view_profile():
    print("Profile opened")


view_profile()
```

```python
#Implement memoization using a function.

def memoize(func):
    cache = {}

    def wrapper(n):
        if n in cache:
            return cache[n]

        result = func(n)
        cache[n] = result
        return result

    return wrapper


@memoize
def square(n):
    print("Calculating...")
    return n * n


print(square(5))
print(square(5))
```

```python
#Write a function to flatten a nested list.

def flatten(lst):
    result = []

    for item in lst:
        if type(item) == list:
            result.extend(flatten(item))
        else:
            result.append(item)

    return result


data = [1, [2, 3], [4, [5, 6]], 7]

print(flatten(data))
```

```python
#Create a function that retries execution if it fails (retry logic).

def retry(func):
    def wrapper():
        max_attempts = 3

        for attempt in range(max_attempts):
            try:
                return func()

            except Exception:
                print("Failed... Retrying")

        print("Maximum attempts reached")

    return wrapper


@retry
def test():
    x = 10 / 0


test()
```

```python
# Write a function that implements custom map() without using built-in map.

def custom_map(func, lst):
    result = []

    for item in lst:
        result.append(func(item))

    return result


def square(n):
    return n * n


numbers = [1, 2, 3, 4]

print(custom_map(square, numbers))
```

```python
#Create a function pipeline (chaining multiple functions dynamically).

def pipeline(value, functions):
    for func in functions:
        value = func(value)

    return value


def add_two(x):
    return x + 2

def multiply_three(x):
    return x * 3

def square(x):
    return x * x


result = pipeline(5, [add_two, multiply_three, square])

print(result)
```
