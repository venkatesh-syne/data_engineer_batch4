```python
import csv 
import json
import pymysql
```

```python
connection = pymysql.connect(
host= 'localhost',user='root', password='Gotham*1998', database='gowthamdb'


)
```

```python
## Reading data from database and creating as a csv file
try:
    with connection.cursor() as cursor:
        cursor.execute('''select * from employee''')
        results = cursor.fetchall()
        with open("emp.csv","w",newline = "") as f:
            write = csv.writer(f) 
            write.writerow(["emp_id","emp_name","dept","salary","manager_id","joining_date"])
            write.writerows(results)
        print('data is successfully uploaded')
    
except Exception as e:
    print('database error',e)
finally : 
    if cursor:
        cursor.close()
    if connection:
        connection.close()
    print('connection closed')
    
```

```python
## Reading data from a file amd writing into database
try:
    with connection.cursor() as cursor:
        data = []
        with open("emp.csv",'r') as f:
            reader = csv.reader(f)
            next(reader)
            for row in reader:
                data.append(tuple(row))
        query = ''' insert into employee
        (emp_id,emp_name,dept,salary,manager_id,joining_date)
        values (%s,%s,%s,%s,%s,%s)'''
        cursor.executemany(query,data)
        connection.commit()
        print('rows are inserted')
except Exception as e:
    print('database error',e)
finally : 
    if cursor:
        cursor.close()
    if connection:
        connection.close()
    print('connection closed')
```

```python
## Reading data from database and creating as a json file
try:
    with connection.cursor() as cursor:
        cursor.execute('''select emp_id,emp_name,dept,salary,manager_id from employee''')
        ##results = cursor.fetchall()
        ##print(results)

        columns = [desc[0] for desc in cursor.description]
        print(columns)

        rows = cursor.fetchall()
        data = [dict(zip(columns, row)) for row in rows]

        with open("emp.json", "w") as f:
            json.dump(data, f, indent=1)

    
except Exception as e:
    print('database error',e)
finally : 
    if cursor:
        cursor.close()
    if connection:
        connection.close()
    print('connection closed')
```
