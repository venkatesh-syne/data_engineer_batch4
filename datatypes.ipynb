## strings 

```python
# Reverse a string.
str = 'abc'
print(str[::-1])
```

```python
# Count vowels in a string.

str = 'vowelsa'
count = 0
for s in str:
    if s in ['a','e','i','o','u']:
        count+=1
    else: continue
print(f'{count}')
```

```python
# Convert a string to uppercase.
str = 'abc'
print(str.upper())
```

```python
# Check whether a string is palindrome.
str = 'madam'
if str == str[::-1]:
    print(f'{str} is palindrome')
else: print('its not a palindrome')
```

```python
# Replace one word with another in a sentence
sen = 'once upon a time'
sen = sen.replace('once','twice')
print(sen)
```

```python
# Find length of a string.
sen = 'abcd'
print(len(sen))
```

```python
# Count how many times a character appears.
sen = 'once upon a time'
char = 'a'
count = 0
for s in sen:
    if s == char:
        count+=1
    else: continue
print(f'{char} is appeared in {count} time')
```

```python
# Split a sentence into words.
sen = 'once upon a time'
words = sen.split()
print(words)
```

## Lists

```python
## Add an element to a list.
lst = [1,2,3,4,5]
lst.append(6)
print(lst)
```

```python
## Remove an element from a list.
lst = [1,2,3,4,5]
lst.remove(2)
print(lst)
```

```python
## Find the number in a list.
lst = [1,2,8,4,3]
print(max(lst))
print(min(lst))
print(sum(lst))
lst.sort()
print(lst)
```

```python
# Loop through a list and print elements.
lst = [1,2,3,4,5]
for a in lst:
    print(a)
```

```python
# Check whether an item exists in a list.
lst = [1,2,3,4,5]
n = 22
if n in lst:
    print('exists')
else: print('not exists')
```

## Dictionary

```python
# Access a value using key.
d1 = {'name': 'abc','age':28}
print(d1)
print(type(d1))
print(d1['age'])
d2=d1.keys() # to get keys
print(d2)
#adding new key value
d1['place']='bangalore'
print(d1)
#update a dict value
d1['age'] = 29
print(d1)
#delete
del d1['place']
print(d1)
#Loop through dictionary keys and values
for key,value in d1.items():
    print(f'key:{key},val:{value}')
# Check if a key exists.
if 'name' in d1:
    print('exists')
#Print all keys from dictionary.
for key in d1:
    print(f'key:{key}')
```
