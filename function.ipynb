```python
#Function to return square of a number 
num = int(input("Enter any number"))
def square(num):
    return num*num
print(square(num))
```

```python
#Function to check even or odd
def check(num):
    if num % 2 == 0:
        return "Even number"
    else:
        return "Odd number"
print(check(7))

```

```python
#Length of string without len()
s = str(input("Enter any string:"))
def length_string(s):
    count = 0 
    for i in s:
        count += 1
    return count
print (length_string(s))
```

```python
#Print elements in list
def print_list(lst):
    for i in lst:
        print(i)
print_list([10,20,30,40])
```

```python
#Palindrome 
def palindrome(text):
    return text == text[::-1]
print(palindrome("radar"))
```

```python
#Factorial
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact 
print (factorial(3))
```

```python
#Count Vowels
def vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for ch in text:
        if ch in vowels:
            count += 1
    return count
print(vowels("HELLOWORLD"))
```

```python
#Largest elements in list
def largest(lst):
    max_num = lst[0]
    for i in lst:
        if i > max_num:
            max_num = i
    return max_num
print(largest([10,13,5,4,29]))
```

```python
#Recursive factorial
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n-1)
print (factorial(4))
```

```python
# sum of args
def total(*args):
    return sum(args)
print(total(1,2,3,4))
```

```python
#Fibonacci series
def fibonacci(n):
    a = 0
    b = 1
    for i in range(n):
        print(a,end= " ")
        c= a + b
        a = b
        b = c
fibonacci(8)
```

```python
#Lambda square function
square = lambda n: n * n
print(square(6))
```

```python
#map() square list
nums = [1,2,3,4,5]
result = list(map(lambda x: x*x, nums))
print(result)
```
