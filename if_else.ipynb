```python
## Check whether a number is positive, negative, or zero
n = int(input())
if n > 0:
    print('num is positive')
elif n < 0:
    print('num is negative')
else:
    print('num is zero')
```

```python
## Check whether a number is even or odd.
n = 4
if n%2 == 0:
    print('num is even')
else:
    print('num is odd')
```

```python
# Find the largest among three numbers
n1 = 32
n2 = 6
n3 = 9
if n1 > n2 and n1 > n3:
    print(f'{n1} is greater')
elif n2 > n1 and n2 > n3:
    print(f'{n2} is greater')
else: print(f'{n3} is greater')
```

```python
# Check whether a year is leap year.
year = 2023
if year % 4 == 0 and year % 100 != 0:
    if year % 400 == 0:
        print(f'{year} is leap year')
    else: print(f'{year} is not a leap year')
else:
    print(f'{year} is not a leap year')

```

```python
## Check eligibility for voting based on age.
age = 17
if age >= 18:
    print('person is eligible to vote')
else: print('person is minor')
```

```python
# Assign grades based on marks using if-elif-else.
marks = 11
if marks > 90 :
    print('A grade')
elif marks > 70 and marks < 90:
    print('B grade')
elif marks > 35 and marks < 70:
    print('Just pass')
else: print('student is failed') 
```

```python
# Check whether a character is vowel or consonant.
def char(ch):
    if ch == 'a' or ch == 'e' or ch == 'i' or ch =='o' or ch == 'u':
        print(f'{ch} is vowel')
    else: print(f'{ch} is consonant')

char1 = 'b'
char(char1)
```

```python
# Create a simple calculator using if-else.
n1 = 10
n2 = 5
op = 'mul'
if op == 'sum':
    print(f'sum is' ,n1+n2)
elif op == 'dif':
    print(f'diff is' ,n1-n2)
elif op == 'mul':
    print(f'mul is' ,n1*n2)
elif op == 'div':
    print(f'div is' ,n1/n2)
else:
    print('invalid operator')
```
