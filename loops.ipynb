```python
# Find factorial of a number using for loop.
def fact(n):
    i = 1
    res = 1

    while (i <= n):
        res = res * i
        i = i + 1
    return res

num = 5
print(fact(num))


```

```python
# Count vowels in a string using loop.
str = 'vowels'
ct = 0
for s in str:
    if s in ['a','e','i','o','u']:
        ct +=1
print(ct)
```

```python
# Find sum of digits of a number and Reverse a number using while loop.

num = 1234
tot = 0
rev = 0
while num > 0:
    tot += num % 10
    rev = rev * 10 + (num % 10)
    num = num // 10
print(tot)
print(f'rev is {rev}')

# palindrome
if num == rev:
    print('its a palindrome')
else: print('its not a palindrome')
```
