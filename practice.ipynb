```python
## sum of num with for loop
a = int(input("enter num"))
sum=0
for i in range(a+1):
    sum =sum+i
print(sum)
```

```python
## sum of n num with while loop
n= int(input())
i=0
sum= 0
while i <= n:
    sum+=i
    i+=1
print(sum)
```
